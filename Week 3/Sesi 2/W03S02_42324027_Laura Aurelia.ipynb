**Simple Linear Regression**

```python
import pandas as pd
df = pd.read_csv('Salary_Data.csv')
print(df.head())
```

Menentukan predictor (X) dan target (y) dari DataFrame untuk Simple Linear Regression

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, 1].values
```

```python
import matplotlib.pyplot as plt
plt.scatter(X, y, color = "red")
plt.xlabel('Years Experience')
plt.ylabel('Salary')
plt.show()
```

```python
#bagi dataset jadi train dan test 80:20
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

Melatih model regresi linear sederhana pada data X_train (variabel independen) dan y_train (target) menggunakan fungsi LinearRegression(). Setelah model dilatih, ia akan menyesuaikan garis regresi terbaik untuk memprediksi nilai target berdasarkan variabel input

```python
# Fitting Simple Linear Regression to the Training set
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

- Menghitung nilai *intercept a*, yaitu nilai *y* saat *x = 0* dalam persamaan *ŷ = a + bx*
- Menghitung *slope* atau koefisien kemiringan *b* yang menunjukkan seberapa besar perubahan *y* saat *x* bertambah

```python
a = regressor.intercept_
b = regressor.coef_[0]

print("Estimated model intercept, a:", a)
print("Estimated model slope, b:", b)
print(f"ŷ = {a:.2f} + {b:.2f}x")
```

```python
# memprediksi hasil pada data uji (Test set) menggunakan model regresi yang telah dilatih
y_pred = regressor.predict(X_test)
y_pred
```

```python
# Menghitung nilai R-Square, yang mengukur seberapa baik model regresi linear
# Semakin mendekati 1, semakin baik model

from sklearn.metrics import r2_score
score = r2_score(y_test, y_pred)
score
```

```python
# Visualizing the Training set results
plt.scatter(X_train, y_train, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
```

```python
# Visualizing the Test set results
plt.scatter(X_test, y_test, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
plt.title('Salary vs Experience (Test set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

**Multiple Linear Regression**

```python
df2 = pd.read_csv('multiple_linear_regression_dataset.csv')
print(df2.head())
```

Menentukan predictor (X) dan target (y) dari DataFrame untuk Multiple Linear Regression

```python
X2 = df2.iloc[:, :-1]
y2 = df2.iloc[:, 2]
```

```python
# Bagi dataset menjadi train dan test 80:20

X_train2, X_test2, y_train2, y_test2 = train_test_split(X2, y2, test_size = 0.2, random_state = 0)
```

Multiple Linear Regression juga menggunakan fungsi yang sama seperti Simple Linear Regression, yaitu LinearRegression() dari Scikit-Learn.

```python
regressor = LinearRegression()
regressor.fit(X_train2, y_train2)
```

```python
# Predicting the Test set results
y_pred2 = regressor.predict(X_test2)
print(y_pred2)
```

- Menghitung nilai *intercept a*, yaitu nilai *y* saat *x = 0* dalam persamaan *ŷ = a + bx*
- Menghitung *slope* atau koefisien kemiringan *b* yang menunjukkan seberapa besar perubahan *y* saat *x* bertambah
- Karena pada Multiple Linear Regression ada lebih dari satu predictor, maka koefisien ditentukan sesuai dengan jumlah predictor

```python
a_ = regressor.intercept_
b1 = regressor.coef_[0]
b2 = regressor.coef_[1]

print("Estimated model intercept, a:", a_)
print("Estimated model slope, b1:", b1)
print("Estimated model slope, b2:", b2)
print(f"ŷ = {a_:.2f} {b1:.2f}x1 + {b2:.2f}x2")
```

```python
# Menghitung nilai R-Square, yang mengukur seberapa baik model regresi linear
# Semakin mendekati 1, semakin baik model

score2 = r2_score(y_test2, y_pred2)
score2
```

## Tugas 1 – Simple Linear Regression
Memprediksi score (y) berdasarkan waktu belajar (X)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

df3 = pd.read_csv('rounded_hours_student_scores.csv')
print(df3.head())
print(df3.shape)
```

```python
df3.info()
print(df3.isnull().sum())
df3.describe()
```

```python
X3 = df3.iloc[:, :-1].values
y3 = df3.iloc[:, -1].values
print('Predictor :', df3.columns[0])
print('Target    :', df3.columns[-1])
print('Korelasi  :', df3.corr().iloc[0, 1].round(4))
```

```python
plt.scatter(X3, y3, color='red')
plt.xlabel('Waktu Belajar (Hours)')
plt.ylabel('Score')
plt.title('Score vs Waktu Belajar')
plt.show()
```

```python
X_train3, X_test3, y_train3, y_test3 = train_test_split(X3, y3, test_size=0.2, random_state=0)
print(X_train3.shape, X_test3.shape)
```

```python
regressor3 = LinearRegression()
regressor3.fit(X_train3, y_train3)
```

```python
a3 = regressor3.intercept_
b3 = regressor3.coef_[0]
print("Estimated model intercept, a:", a3)
print("Estimated model slope, b:", b3)
print(f"ŷ = {a3:.2f} + {b3:.2f}x")
```

```python
y_pred3 = regressor3.predict(X_test3)
pd.DataFrame({'Aktual': y_test3, 'Prediksi': y_pred3.round(2)})
```

```python
score3 = r2_score(y_test3, y_pred3)
print('R-Square :', score3)
print('MAE      :', mean_absolute_error(y_test3, y_pred3))
print('RMSE     :', np.sqrt(mean_squared_error(y_test3, y_pred3)))
```

```python
plt.scatter(X_train3, y_train3, color='red')
plt.plot(X_train3, regressor3.predict(X_train3), color='blue')
plt.title('Score vs Waktu Belajar (Training set)')
plt.xlabel('Waktu Belajar'); plt.ylabel('Score')
plt.show()
```

```python
plt.scatter(X_test3, y_test3, color='red')
plt.plot(X_train3, regressor3.predict(X_train3), color='blue')
plt.title('Score vs Waktu Belajar (Test set)')
plt.xlabel('Waktu Belajar'); plt.ylabel('Score')
plt.show()
```

```python
print('Prediksi score untuk 7 jam belajar:', regressor3.predict([[7]])[0])
```

## Tugas 2 – Multiple Linear Regression
Memprediksi sales (y) berdasarkan biaya marketing di youtube, facebook, dan newspaper

```python
df4 = pd.read_csv('Marketing_Data.csv')
df4 = df4.loc[:, ~df4.columns.str.contains('^Unnamed')]
print(df4.head())
print(df4.shape)
```

```python
df4.info()
print(df4.isnull().sum())
df4.describe()
```

```python
print(df4.corr())
```

```python
X4 = df4.iloc[:, :-1]
y4 = df4.iloc[:, -1]
print('Predictor:', list(X4.columns))
print('Target   :', y4.name)

X_train4, X_test4, y_train4, y_test4 = train_test_split(X4, y4, test_size=0.2, random_state=0)
print(X_train4.shape, X_test4.shape)
```

```python
X_train4, X_test4, y_train4, y_test4 = train_test_split(X4, y4, test_size=0.2, random_state=0)
print(X_train4.shape, X_test4.shape)
```

```python
regressor4 = LinearRegression()
regressor4.fit(X_train4, y_train4)
```

```python
y_pred4 = regressor4.predict(X_test4)
pd.DataFrame({'Aktual': y_test4.values, 'Prediksi': y_pred4.round(2)}).head(10)
```

```python
a4 = regressor4.intercept_
b1_4, b2_4, b3_4 = regressor4.coef_

print("Estimated model intercept, a:", a4)
for nama, koef in zip(X4.columns, regressor4.coef_):
    print(f"Estimated model slope ({nama}):", koef)
print(f"ŷ = {a4:.4f} + ({b1_4:.4f})x1 + ({b2_4:.4f})x2 + ({b3_4:.4f})x3")
print(f"x1 = {X4.columns[0]}, x2 = {X4.columns[1]}, x3 = {X4.columns[2]}")
```

```python
score4 = r2_score(y_test4, y_pred4)
print('R-Square :', score4)
print('MAE      :', mean_absolute_error(y_test4, y_pred4))
print('RMSE     :', np.sqrt(mean_squared_error(y_test4, y_pred4)))
```

```python
plt.scatter(y_test4, y_pred4, color='green')
batas = [min(y_test4.min(), y_pred4.min()), max(y_test4.max(), y_pred4.max())]
plt.plot(batas, batas, 'r--')
plt.xlabel('Sales Aktual'); plt.ylabel('Sales Prediksi')
plt.title('Aktual vs Prediksi (Test set)')
plt.show()
```
