2. temperature_fahrenheit vs air_quality_Sulphur_dioxide = -0.023152

- Meskipun tandanya negatif (artinya arahnya sedikit berlawanan), kekuatannya terlalu lemah untuk dianggap sebagai pola yang nyata. Nilai sekecil ini bisa saja muncul murni karena variasi acak dalam data, bukan hubungan sebab-akibat.

3. pressure_mb vs air_quality_Sulphur_dioxide = 0.018620
- Sama seperti poin sebelumnya, nilai ini sangat mendekati 0, menunjukkan tidak ada hubungan linear yang berarti antara tekanan udara dan kadar Sulphur Dioxide.

```python
print(df['Previous Purchases'].describe())
```

### Analisa Statistik Deskriptif: Previous Purchases

Hasil yang keluar:

| Statistik | Nilai |
|---|---|
| count | 3900 |
| mean | 25.351538 |
| std | 14.447125 |
| min | 1.0 |
| 25% (Q1) | 13.0 |
| 50% (Q2 / median) | 25.0 |
| 75% (Q3) | 38.0 |
| max | 50.0 |

Penjelasan detail:

- Mean (25.35): rata-rata, setiap pelanggan dalam dataset ini sudah pernah melakukan pembelian sebanyak 25 kali sebelumnya. Ini menunjukkan basis pelanggan yang dianalisis termasuk pelanggan yang cukup aktif/loyal secara keseluruhan.

- Std (14.45): standar deviasi ini termasuk besar dibandingkan rata-ratanya sendiri (lebih dari setengah nilai mean). Ini berarti jumlah pembelian sebelumnya antar pelanggan sangat bervariasi, ada pelanggan yang benar-benar baru, ada juga yang sudah sangat sering belanja.

- Min (1) dan Max (50): rentang datanya terbentang dari pelanggan yang baru pertama kali tercatat (1 kali pembelian) sampai pelanggan paling loyal (50 kali pembelian). Rentang sejauh ini (49 poin) menunjukkan populasi pelanggan yang beragam, dari yang baru sampai yang sangat setia.

- Q1 (13): 25% pelanggan dengan riwayat pembelian paling sedikit, jumlah pembelian sebelumnya mereka ada di bawah 13 kali. Ini adalah kelompok pelanggan yang relatif baru.

- Q2 / median (25): setengah dari seluruh pelanggan memiliki riwayat pembelian di bawah 25 kali, dan setengah lagi di atas 25 kali. Titik tengah data.

- Q3 (38): 75% pelanggan memiliki riwayat pembelian di bawah 38 kali, sehingga 25% sisanya (pelanggan paling loyal) sudah membeli lebih dari 38 kali.

- Kesimpulan sebaran data: karena nilai median (25) sangat dekat dengan mean (25.35), dan jarak Q1 ke Q2 (13 ke 25 = 12 poin) hampir sama dengan jarak Q2 ke Q3 (25 ke 38 = 13 poin), data ini tersebar secara cukup merata dan simetris, tidak menumpuk di satu sisi (tidak skewed). Artinya jumlah pelanggan baru, menengah, dan sangat loyal proporsinya relatif seimbang dalam dataset ini.

```python
korelasi_age = df[['Age', 'Purchase Amount (USD)']].corr()
print(korelasi_age)
```

### Analisa Korelasi: Age vs Purchase Amount (USD)

Hasil korelasi: -0.010424

Penjelasan detail:

- Nilai korelasi ini sangat mendekati 0 (skala korelasi berkisar dari -1 sampai 1), sehingga termasuk kategori korelasi yang dapat diabaikan / tidak berarti secara praktis.
- Tandanya negatif, tapi karena besarnya kurang dari 0.1, tanda ini tidak punya makna praktis, bisa jadi hanya fluktuasi acak dalam data, bukan pola sungguhan.
- Artinya: usia pelanggan tidak bisa dipakai untuk memprediksi seberapa besar nominal uang yang mereka belanjakan. Pelanggan muda dan pelanggan tua memiliki kecenderungan pola belanja (dalam nominal USD) yang relatif sama, tidak ada kelompok usia tertentu yang secara konsisten belanja lebih besar atau lebih kecil.
- Ini masuk akal jika dataset ini bersifat sintetis/acak (seperti disebutkan di deskripsi Kaggle-nya sebagai "Synthetic Data"), di mana nominal pembelian memang sengaja tidak dikaitkan dengan usia pelanggan.

```python
korelasi_rating = df[['Purchase Amount (USD)', 'Review Rating']].corr()
print(korelasi_rating)
```

### Analisa Korelasi: Purchase Amount (USD) vs Review Rating

Hasil korelasi: 0.030776

Penjelasan detail:

- Sama seperti korelasi sebelumnya, nilai ini sangat mendekati 0, sehingga termasuk kategori korelasi yang dapat diabaikan.
- Kali ini tandanya positif, tapi karena nilainya jauh di bawah 0.1, ini tidak cukup kuat untuk disebut sebagai hubungan yang nyata antar dua variabel.
- Artinya: besar-kecilnya nominal uang yang dibelanjakan pelanggan tidak berhubungan dengan rating ulasan (review rating) yang mereka berikan. 
  - Pelanggan yang belanja dalam jumlah besar belum tentu memberikan rating yang tinggi.
  - Pelanggan yang memberikan rating tinggi belum tentu juga adalah pelanggan yang belanja dalam jumlah besar.
- Ini menunjukkan bahwa kepuasan pelanggan (diwakili oleh review rating) kemungkinan lebih dipengaruhi oleh faktor lain seperti kualitas produk, pelayanan, atau ekspektasi pribadi, bukan oleh seberapa mahal barang yang mereka beli.

### Kesimpulan Akhir Tugas Nomor 2

Dari kedua analisis korelasi di atas, dapat disimpulkan bahwa nominal pembelian (Purchase Amount) pada dataset ini tidak dipengaruhi secara berarti oleh usia pelanggan maupun rating ulasan yang diberikan. Kedua korelasi menunjukkan angka yang sangat mendekati nol, yang berarti tidak ada hubungan linear yang bisa diandalkan antar variabel-variabel tersebut.
